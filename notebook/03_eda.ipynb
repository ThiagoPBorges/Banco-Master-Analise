```python
import pandas as pd
import os
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

# Criação de novas Features

```python
# --- RENTABILIDADE ---
# ROE (Return on Equity)
df_analitico['kpi_roe'] = df_analitico['lucro_liquido'] / df_analitico['patrimonio_liquido']
# ROA (Return on Asset)
df_analitico['kpi_roa'] = df_analitico['lucro_liquido'] / df_analitico['ativo_total']

# --- QUALIDADE ---
# Dependência de "Outras Receitas".
# Se o banco depende muito de receitas não-operacionais, isso é não recorrente.
# Fórmula: Outras Receitas / (Total de Receitas Operacionais)
df_analitico['receita_total_estimada'] = df_analitico['resultado_intermediacao'] + df_analitico['outras_receitas_despesas']

df_analitico['kpi_qualidade_lucro'] = (df_analitico['outras_receitas_despesas'] + df_analitico['Resultado de Participações (t)']) / df_analitico['receita_total_estimada']

# --- VOLUMETRIA ---
# Tamanho do Banco (Log) para separar Pequenos de Gigantes
df_analitico['tamanho_log'] = np.log10(df_analitico['ativo_total'])

# Taxa média (Yield)
df_analitico['yield_carteira'] = df_analitico['Rendas de Operações de Crédito (c)'] / df_analitico['carteira_credito']
df_analitico['yield_carteira'] = df_analitico['yield_carteira'].replace([np.inf, -np.inf], np.nan)

df_analitico['custo_funding'] = (df_analitico['Despesa_Captacao_Total'] * -1) / df_analitico['captacoes']
# Custo Implícito = Quanto pagou de juros / Quanto tinha de dívida
df_analitico['taxa_funding_implicita'] = df_analitico['custo_funding'] / df_analitico['captacoes']

df_analitico.info()
```

# Custo de Funding: Comparação com o mercado

## O Benchmark de Eficiência (Linha Preta)
A linha tracejada preta representa a taxa de **CDI**.
* **Bancos Sólidos (Linhas Cinzas):** Note que a grande massa de bancos operam frequentemente **abaixo ou colada** à linha do CDI. Isso indica alta confiança do mercado e capacidade de captar recursos.

## A Ineficiência do Banco Master (Linha Azul)
O Banco Master apresenta um comportamento divergente da média:
* **Prêmio de Risco:** A linha azul opera consistentemente **acima do CDI**. Diferente dos pares que conseguem captar a 95% ou 100% do CDI, o Master frequentemente paga taxas superiores à Selic.
* **Picos:** É possível observar momentos (como em 2023) onde o custo dispara, abrindo um spread significativo contra o mercado.

## Conclusão de Negócio
Este gráfico prova que o dinheiro "custa mais caro" para o Master do que para seus concorrentes.
* **Impacto na Margem:** Se o banco paga mais caro na entrada, ele entra no jogo do crédito com uma desvantagem competitiva.
* **Obrigatoriedade do Risco:** Para manter a margem de lucro positiva pagando esse ágio no funding, a instituição não tem escolha a não ser operar em linhas de crédito de juros altíssimos (High Yield), que são inerentemente mais arriscadas. O funding caro é a raiz da estratégia agressiva do banco.

```python
# ... (sua parte de ordenação continua igual)
df_analitico = df_analitico.sort_values(['instituicao', 'data_base'])

# 1. LÓGICA PARA DESPESA (FLUXO) - Mantém a sua lógica de desacumular
df_analitico['Despesa_Trimestre'] = df_analitico['Despesa_Captacao_Total']
mascara_acumulada = df_analitico['data_base'].dt.month.isin([6, 12])

# Se for junho ou dezembro, subtrai o anterior para pegar só o trimestre
df_analitico.loc[mascara_acumulada, 'Despesa_Trimestre'] = (
    df_analitico['Despesa_Captacao_Total'] - df_analitico['Despesa_Captacao_Total'].shift(1)
)

# 2. LÓGICA PARA CAPTAÇÕES (ESTOQUE) - Aqui mudamos para MÉDIA
# Não subtraímos. Nós pegamos a média entre o saldo atual e o anterior
# Isso suaviza picos de entrada de dinheiro no último dia do mês
df_analitico['captacoes_media'] = (
    df_analitico['captacoes'] + df_analitico['captacoes'].shift(1)
) / 2

# Tratamento para o primeiro trimestre de cada banco (onde o shift(1) pega outro banco ou NaN)
# Se mudou o banco, não podemos usar a média com o anterior. Usamos o saldo final mesmo.
mask_mudou_banco = df_analitico['instituicao'] != df_analitico['instituicao'].shift(1)
df_analitico.loc[mask_mudou_banco, 'captacoes_media'] = df_analitico['captacoes']


# 3. CÁLCULO DA TAXA
# Usamos a Despesa do Trimestre dividida pela Captação Média
# (Multipliquei por -1 assumindo que sua despesa está negativa no banco de dados)
df_analitico['custo_funding_tri_real'] = (df_analitico['Despesa_Trimestre'] * -1) / df_analitico['captacoes_media']

# Anualização continua igual
df_analitico['custo_funding_ano_real'] = ((1 + df_analitico['custo_funding_tri_real']) ** 4 - 1)

cdi_historico = {
    '2021Q1': 0.019, '2021Q2': 0.032, '2021Q3': 0.050, '2021Q4': 0.077,
    '2022Q1': 0.101, '2022Q2': 0.121, '2022Q3': 0.139, '2022Q4': 0.134,
    '2023Q1': 0.136, '2023Q2': 0.132, '2023Q3': 0.135, '2023Q4': 0.118,
    '2024Q1': 0.109, '2024Q2': 0.105, '2024Q3': 0.110, '2024Q4': 0.111,
    '2025Q1': 0.125, '2025Q2': 0.140, '2025Q3': 0.157, '2025Q4': 0.109
}
# Mapear o CDI para dentro do df_bancos
df_bancos['taxa_cdi'] = df_bancos['Trimestre'].map(cdi_historico)

paleta_cores = {banco: "#2E48DA" if banco == banco_foco else '#d9d9d9' for banco in todos_bancos}
espessuras = {banco: 4.0 if banco == banco_foco else 1.5 for banco in todos_bancos}

plt.figure(figsize=(15,8))

sns.lineplot(
    data=df_bancos,
    x='Trimestre',
    y='custo_funding_ano_real',
    hue='instituicao',
    size='instituicao',
    palette=paleta_cores,
    sizes=espessuras
)

sns.lineplot(
    data=df_bancos, 
    x='Trimestre', 
    y='taxa_cdi', 
    color='black', 
    linestyle='--', 
    linewidth=2.5, 
    label='CDI (Ref. Mercado)',
    errorbar=None
)

ax.yaxis.set_major_formatter(ticker.PercentFormatter(1.0))

plt.title('Custo de Funding Bancos vs CDI (% a.a.)', fontsize=16, fontweight='bold', loc='left', x=0.25)
plt.ylabel('Taxa Anualizada (% a.a.)')
plt.xlabel('Trimestre')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()
```
