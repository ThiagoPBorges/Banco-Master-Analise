```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd

from selenium.webdriver.common.by import By          # Serve para achar coisas (pelo ID, Nome, Link)
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import os
import glob # Biblioteca para encontrar arquivos na pasta

import shutil
```

```python
options = webdriver.ChromeOptions()

servico = Service(ChromeDriverManager().install())
navegador = webdriver.Chrome(service=servico)

navegador.get("https://www3.bcb.gov.br/ifdata/")

navegador.maximize_window()
```

```python
txt_data = "06/2025"
txt_data_antiga = 'Dados de 2000 a 2024'
btn_data = 'btnDataBase'
data_d = pd.to_datetime(txt_data, format='%m/%Y', errors='coerce')

txt_tipo_instituicao = 'Instituições Individuais'
btn_tipo_instituicao = 'btnTipoInst'

txt_tipo_relatorio = 'Resumo'
btn_tipo_relatorio = 'btnRelatorio'
```

```python
# Se for antes ou 2024, o código clica na seção onde estão essas informações antigas.
if data_d.year <= 2024:

    # Troca de aba se for ano antigo
    selecao_data_antiga = navegador.find_element(By.LINK_TEXT, txt_data_antiga)
    selecao_data_antiga.click()

# Clica em data-base
botao_data = navegador.find_element(By.ID, btn_data)
botao_data.click()
print("Lista de datas aberta")

# Se for antigo, esperar mais tempo, pois tem mais informações
if data_d.year <= 2024:
    time.sleep(15)
else:
    time.sleep(0.5)

# Clica para escolher a data
escolha_data = navegador.find_element(By.LINK_TEXT, txt_data)
escolha_data.click()
print("Data selecionada")
time.sleep(0.5)

# Clica no Tipo de Instituição
botao_tipo_instituicao = navegador.find_element(By.ID, btn_tipo_instituicao)
botao_tipo_instituicao.click()
print("Botão instituição selecionado")
time.sleep(0.5)

# Clica para escolher o Tipo de Instituição
escolha_tipo_instituicao = navegador.find_element(By.LINK_TEXT, txt_tipo_instituicao)
escolha_tipo_instituicao.click()
print("Escolha instituição feita")
time.sleep(0.5)

# Clica em Tipo de Relatório
botao_tipo_relatorio = navegador.find_element(By.ID, 'btnRelatorio')
botao_tipo_relatorio.click()
print("Botão Relatório selecionado")
time.sleep(0.5)

# Clica para escolher o Tipo de Relatório
botao_tipo_relatorio = navegador.find_element(By.LINK_TEXT, txt_tipo_relatorio)
botao_tipo_relatorio.click()
print("Escolha Relatório feita")
```

Exportar e Salvar

```python
print("Aguardando botão CSV...")
esperar = WebDriverWait(navegador, 20)
exportar_csv = esperar.until(EC.element_to_be_clickable((By.ID, 'aExportCsv')))
exportar_csv.click()
print("Download iniciado na pasta Downloads padrão")

# Espera o download acabar
time.sleep(1) 

# BUSCAR O ARQUIVO ONDE ELE ESTIVER

# Definir onde é a pasta Downloads do seu computador
# Encontra a pasta do usuário (ex: C:\Users\Thiag) e adiciona "Downloads"
pasta_downloads_padrao = os.path.join(os.path.expanduser("~"), "Downloads")

# Definir o nome novo e onde queremos salvar (na pasta do projeto)
nome_final = f"resumo_{data_d.strftime('%Y%m')}.csv"
caminho_final_projeto = os.path.join('../dados/if_data_bacen_exportacao', nome_final)

# Procurar o arquivo mais recente NA PASTA DOWNLOADS
print(f"Procurando arquivo em: {pasta_downloads_padrao}")
lista_csvs = glob.glob(os.path.join(pasta_downloads_padrao, "*.csv"))

if lista_csvs:
    # Pega o arquivo modificado mais recentemente na pasta Downloads
    arquivo_baixado = max(lista_csvs, key=os.path.getctime)
    
    print(f"Arquivo encontrado: {arquivo_baixado}")

    # 4. Mover e Renomear
    try:
        # Se já existe um arquivo com esse nome na sua pasta de projeto, deleta antes
        if os.path.exists(caminho_final_projeto):
            os.remove(caminho_final_projeto)
        
        # Move da pasta Downloads para a pasta do Projeto com o nome novo
        shutil.move(arquivo_baixado, caminho_final_projeto)
        
        print(f"SUCESSO! Arquivo movido e salvo como: {nome_final}")
        print(f"Local: {caminho_final_projeto}")
        
    except Exception as e:
        print(f"Erro ao mover o arquivo: {e}")
else:
    print("ERRO: O arquivo não foi encontrado nem na pasta Downloads.")
```
